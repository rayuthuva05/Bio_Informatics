```python
import pandas as pd
import numpy as np
from collections import Counter
import matplotlib.pyplot as plt
```

```python
DNA="ATGCGTAGCTAAG"
k=5
```

```python
l=len(DNA)
print(l)
motifs=[None]*(l-k+1)
```

```python
j=0
for i in range(l-4):
    motifs[j]=DNA[i:i+5]
    j+=1

print(motifs)
```

```python
motifs_counts=Counter(motifs)
print(motifs_counts)
```

```python
def plot_top_motifs(motifs_counts, top_n=10):
    common=motifs_counts.most_common(top_n)
    labels=[m[0] for m in common]
    values=[m[1] for m in common]

    plt.figure(figsize=(10,5))
    plt.bar(labels,values)
    plt.xlabel('Mofits')
    plt.ylabel('Frequency')
    plt.show()

plot_top_motifs(motifs_counts)
```
