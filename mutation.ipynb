```python
DNA="ATGCCGTAGCTAAGGCTT"
```

```python
#substitution

def substitution_mutation(seq,pos,new_base):

    if pos<0 or pos>=len(seq):
        print('Enrror ')

    else:
        mutated_seq=seq[:pos]+new_base+seq[pos+1:]
    return mutated_seq

print("Original:",DNA)
print('Mututed:',substitution_mutation(DNA,4,'A'))
```

```python
def Insertion(seq,pos,new_base):

    if pos<0 or pos>=len(seq):
        print("warning")

    else:
        mutated_seq=seq[:pos]+new_base+seq[pos:]
    return mutated_seq

print("Original:",DNA)
print("Mutated:",Insertion(DNA,5,'T'))
```

```python
def Deletion(seq,pos):

    if pos<0 or pos>=len(seq):
        print("Warning")
    else:
        mutated_seq=seq[:pos]+seq[pos+1:]
    return mutated_seq

print("Original:",DNA)
print("Mutated:",Deletion(DNA,4))
```

```python
def sequence_identity(seq1, seq2):

    if len(seq1)!=len(seq2):
        raise ValueError("both sequence lengths are should same")
    
    matches=sum(a==b for a,b in zip(seq1,seq2));
    identity=matches/len(seq1)*100
    return identity

def sequence_similarity(seq1, seq2):

    return sequence_identity(seq1, seq2)


# Example
seqA = "AUGGCUACGAAGUAA"
seqB = "AUGGCCACGAAGCAA"

id_percent = sequence_identity(seqA, seqB)
sim_percent = sequence_similarity(seqA, seqB)

print("Sequence A:", seqA)
print("Sequence B:", seqB)
print(f"Identity:  {id_percent:.2f}%")
print(f"Similarity: {sim_percent:.2f}%")
```
