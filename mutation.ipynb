```python
def substitution_mutation(seq, position, new_base):
    
    if position < 0 or position >= len(seq):
        raise ValueError("Position is outside the sequence length.")

    mutated = seq[:position] + new_base + seq[position+1:]
    return mutated

mrna = "AUGGCUACGAAGUAA"
pos = 5
new_nt = "A"

mutated_seq = substitution_mutation(mrna, pos, new_nt)

print("Original:", mrna)
print("Mutated :", mutated_seq)
print(f"Changed position {pos} from {mrna[pos]} → {new_nt}")
```

```python
def insertion_mutation(seq, position, new_base):
    if position < 0 or position > len(seq):
        raise ValueError("Position is outside the sequence length.")

    mutated = seq[:position] + new_base + seq[position:]
    return mutated

mrna = "AUGGCUACGAAGUAA"
pos = 3
new_nt = "U"

mutated_seq = insertion_mutation(mrna, pos, new_nt)

print("Original:", mrna)
print("Mutated :", mutated_seq)
print(f"Inserted '{new_nt}' at position {pos}")
```

```python
def deletion_mutation(seq, position):

    if position < 0 or position >= len(seq):
        raise ValueError("Position is outside the sequence length.")

    mutated = seq[:position] + seq[position+1:]
    return mutated


# Example usage
mrna = "AUGGCUACGAAGUAA"
pos = 4

mutated_seq = deletion_mutation(mrna, pos)

print("Original:", mrna)
print("Mutated :", mutated_seq)
print(f"Deleted nucleotide '{mrna[pos]}' at position {pos}")
```

```python
def sequence_identity(seq1, seq2):
    """
    Calculate percent identity between two sequences.
    Identity = exact character matches.
    """
    if len(seq1) != len(seq2):
        raise ValueError("Sequences must be the same length.")

    matches = sum(a == b for a, b in zip(seq1, seq2))
    identity = matches / len(seq1) * 100
    return identity


def sequence_similarity(seq1, seq2):
    """
    Simple similarity: here we define similarity = identity
    (you can extend this by grouping similar amino acids).
    """
    return sequence_identity(seq1, seq2)


# Example
seqA = "AUGGCUACGAAGUAA"
seqB = "AUGGCCACGAAGCAA"

id_percent = sequence_identity(seqA, seqB)
sim_percent = sequence_similarity(seqA, seqB)

print("Sequence A:", seqA)
print("Sequence B:", seqB)
print(f"Identity:  {id_percent:.2f}%")
print(f"Similarity: {sim_percent:.2f}%")
```
